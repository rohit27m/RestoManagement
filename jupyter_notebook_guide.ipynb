# Jupyter Notebook Getting Started Guide

This notebook will guide you through the basics of installing, opening, and using Jupyter Notebook.

## 1. Install Jupyter Notebook

Jupyter Notebook can be installed using either pip or conda package manager.

```python
# Install using pip
# !pip install notebook

# Or install using conda
# !conda install -c conda-forge notebook

# Check if Jupyter is installed
import sys
print(f"Python version: {sys.version}")
print("\nTo install Jupyter Notebook, run one of the commands above in your terminal.")
```

## 2. Launch Jupyter Notebook from Terminal

Once installed, you can start the Jupyter Notebook server from your command line.

```python
# To launch Jupyter Notebook, open your terminal and run:
# jupyter notebook

# This will:
# 1. Start the Jupyter Notebook server
# 2. Open your default web browser
# 3. Display the Jupyter dashboard at http://localhost:8888

print("Launch command: jupyter notebook")
print("The server will run on: http://localhost:8888")
```

## 3. Navigate the Jupyter Dashboard

The Jupyter Dashboard is a file browser interface where you can:
- Browse directories and files
- Create new notebooks
- Upload files
- Manage running notebooks

## 4. Create a New Notebook

From the Jupyter Dashboard:
1. Click the "New" button in the top right
2. Select "Python 3" (or your preferred kernel)
3. A new notebook will open in a new tab

## 5. Execute Code Cells

Notebooks consist of cells that can contain code or markdown. Let's try executing some Python code:

```python
# Example 1: Simple calculation
result = 10 + 20
print(f"10 + 20 = {result}")
```

```python
# Example 2: Working with lists
numbers = [1, 2, 3, 4, 5]
squared = [n**2 for n in numbers]
print(f"Original: {numbers}")
print(f"Squared: {squared}")
```

```python
# Example 3: Creating a simple function
def greet(name):
    return f"Hello, {name}! Welcome to Jupyter Notebook."

message = greet("User")
print(message)
```

### Cell Execution Tips

- **Run a cell**: Press `Shift + Enter` or click the "Run" button
- **Insert cell above**: Press `A` (in command mode)
- **Insert cell below**: Press `B` (in command mode)
- **Delete cell**: Press `D + D` (press D twice in command mode)
- **Change to markdown**: Press `M` (in command mode)
- **Change to code**: Press `Y` (in command mode)

## 6. Save and Close Notebooks

### Saving Your Work
- Notebooks auto-save periodically
- Manual save: Click the save icon or press `Ctrl + S` (Windows/Linux) or `Cmd + S` (Mac)
- File menu: `File → Save and Checkpoint`

### Closing the Notebook
1. Close the browser tab with the notebook
2. From the dashboard, select the notebook and click "Shutdown"
3. To stop the Jupyter server, press `Ctrl + C` in the terminal where it's running

```python
# You can also programmatically check notebook information
import datetime

print(f"Notebook completed on: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
print("\nHappy coding with Jupyter Notebook! 🚀")
```
